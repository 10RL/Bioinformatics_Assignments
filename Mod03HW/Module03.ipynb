```r
#Problem 6: Counting DNA Nucleotides
## Used chat to correct errors with this section "for (letter in strsplit(dna_string, "")[[1]]){""
dna_string <- readLines("/content/rosalind_dna.txt")
A<-0
C<-0
T<-0
G<-0 #Creating the counters for each
#Loop through each nucleotide and count it
for (letter in strsplit(dna_string, "")[[1]]){ #Looping individually through each letter. Start with a base command "if" then use "else" for other parameters that must still follow the same command
  if (letter == "A") {
    A <- A+1
  } else if (letter == "C") {
    C <- C+1
  }else if (letter == "T") {
    T <- T+1
  }else if (letter == "G") {
    G <- G+1
  }
}

cat(A,C,G,T, "\n")

```

```r
#Problem 8: Translating RNA to Protein
##used chat to learn how to separate out individual codons and how to join the string back together (I had to look up paste0).
rna_string <- readLines("/content/rosalind_prot.txt")
codon_table <- c( #codons from table provided in Rosiland
  "UUU"="F",      "CUU"="L",      "AUU"="I",      "GUU"="V",
  "UUC"="F",      "CUC"="L",      "AUC"="I",      "GUC"="V",
  "UUA"="L",      "CUA"="L",      "AUA"="I",      "GUA"="V",
  "UUG"="L",      "CUG"="L",      "AUG"="M",      "GUG"="V",
  "UCU"="S",      "CCU"="P",      "ACU"="T",      "GCU"="A",
  "UCC"="S",      "CCC"="P",      "ACC"="T",      "GCC"="A",
  "UCA"="S",      "CCA"="P",      "ACA"="T",      "GCA"="A",
  "UCG"="S",      "CCG"="P",      "ACG"="T",      "GCG"="A",
  "UAU"="Y",      "CAU"="H",      "AAU"="N",      "GAU"="D",
  "UAC"="Y",      "CAC"="H",      "AAC"="N",      "GAC"="D",
  "UAA"="Stop",   "CAA"="Q",      "AAA"="K",      "GAA"="E",
  "UAG"="Stop",   "CAG"="Q",      "AAG"="K",      "GAG"="E",
  "UGU"="C",      "CGU"="R",      "AGU"="S",      "GGU"="G",
  "UGC"="C",      "CGC"="R",      "AGC"="S",      "GGC"="G",
  "UGA"="Stop",   "CGA"="R",      "AGA"="R",      "GGA"="G",
  "UGG"="W",      "CGG"="R",      "AGG"="R",      "GGG"="G"
)
protein <- ""
for(i in seq(1, nchar(rna_string), by = 3)) {  #reads the sequence by three. Had to look up nchar- it counts the number of characters in a string
  individual_codon <- substr (rna_string, i, i+2) #this tells the code to start at position i, collect characters i through i+2, than that three character thing becomes a individual_codon
  if (codon_table[individual_codon]=="Stop") { #stop the code if a stop codon is reached
    break
  }
  protein <- paste0(protein, codon_table[individual_codon]) #paste0 combines strings
}

cat(protein, "\n")
```

```r
#Question 9: Computing GC Content
##Chat used to correct errors

dna_string9 <-readLines("/content/rosalind_gc.txt")
#The below section allows me set up blank paramaters from which I can build the rest of my code
current_id <- ""
current_sequence <- ""
highest_id <- ""
highest_gc <- 0 #Starts with the percentage at zero

#The below section is a loop that identifies IDs, calculates the gc, and adds DNA to the sequence when they are identified
for (line in dna_string9) {
  if (substr(line,1,1) == ">"){ #Checks if the line starts with a sequence ID. The "substr" tool was used in previous code aswell.
    if (current_sequence != "") { # != means "not equal to". this checks that the current sequence does contain information and is not an empty string
      gc <- 0
      for (letter in strsplit(current_sequence, "")[[1]]){
        if (letter == "G" || letter == "C"){
          gc <- gc + 1
        }
      }
  gc_content <- (gc / nchar(current_sequence))*100 #nchar means number of characters. it is checking the total number of bases in the sequence. The full line calculates the gc%
  if (gc_content > highest_gc) {
    highest_gc <- gc_content
    highest_id <- current_id
  }
}

current_id <- substr(line,2,nchar(line))
current_sequence <- ""
}else{
  current_sequence <- paste0(current_sequence, line) #past0 was also used earlier. It recombines the string.
}
}

gc <- 0

#this is another loop that is used to calculate the gc of the final dna sequence
for (letter in strsplit(current_sequence, "")[[1]]){
  if (letter == "G" || letter == "C") {
    gc <-gc +1 #over-writes the original gc so that it can calculate the gc for the final dna
  }
}

gc_content <- (gc / nchar(current_sequence)) * 100 #nchar again used to calculate number of bases
 if (gc_content > highest_gc) {
    highest_gc <- gc_content
    highest_id <- current_id
 }

#Cat is a very specific way to print the data that allows you to more easily control how it is viewed
 cat(highest_id, "\n")
 cat(highest_gc, "\n")
```
